```python

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 1. Load and prepare dataset
df = pd.read_csv("Telco_Customer_Churn_Dataset.csv")
df.columns = df.columns.str.lower().str.replace(' ', '_')
df['totalcharges'] = pd.to_numeric(df['totalcharges'], errors='coerce')
df['totalcharges'] = df['totalcharges'].fillna(df['totalcharges'].median())

# 2. Tenure Segmentation Categories
bins = [-1, 12, 36, 72]
labels = ['0-12 months', '13-36 months', '37+ months']
df['tenure_group'] = pd.cut(df['tenure'], bins=bins, labels=labels)

# --- CHART 1: Interactive Donut Chart (Plotly) ---
tenure_counts = df['tenure_group'].value_counts().reset_index()
tenure_counts.columns = ['tenure_group', 'count']

fig_donut = px.pie(
    tenure_counts,
    values='count',
    names='tenure_group',
    hole=0.4,
    title='<b>Customer Distribution by Tenure Group (Interactive Donut)</b>',
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig_donut.update_traces(textinfo='percent+label', hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}')
fig_donut.show()

# --- CHART 2: Interactive Clustered Bar Chart with Annotations (Plotly) ---
avg_charges = df.groupby(['tenure_group', 'churn'], observed=False)['monthlycharges'].mean().reset_index()

fig_bar = px.bar(
    avg_charges,
    x='tenure_group',
    y='monthlycharges',
    color='churn',
    barmode='group',
    title='<b>Avg Monthly Charges by Tenure & Churn Status (Interactive Clustered Bar)</b>',
    labels={'tenure_group': 'Tenure Category', 'monthlycharges': 'Avg Monthly Charges ($)', 'churn': 'Churn Status'},
    color_discrete_map={'No': '#2ca02c', 'Yes': '#d62728'}
)

# Add interactive trend annotation
fig_bar.add_annotation(
    x='0-12 months',
    y=75,
    text="<b>Higher charges strongly correlate<br>with early churn!</b>",
    showarrow=True,
    arrowhead=2,
    arrowcolor="black",
    arrowsize=1,
    arrowwidth=2,
    ax=40,
    ay=-50,
    bgcolor="yellow",
    opacity=0.8
)

fig_bar.update_layout(hovermode="x unified")
fig_bar.show()
```
